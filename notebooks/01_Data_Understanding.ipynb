Step 14 — Build Attribution Models

I'll calculate four additional models:

First-Touch
Linear
Time-Decay
Position-Based

```python
first_touch = (
    touchpoints
    .sort_values(["user_id", "touch_order"])
    .groupby("user_id")
    .first()
    .reset_index()
)

first_touch_converted = first_touch[
    first_touch["converted"] == 1
].copy()

print(
    "Converted customers:",
    first_touch_converted["user_id"].nunique()
)
```

```python
first_touch_summary = (
    first_touch_converted
    .groupby("channel")
    .agg(
        attributed_conversions=("user_id", "nunique"),
        attributed_revenue=("revenue", "sum")
    )
)

first_touch_summary
```

```python
first_touch_summary = first_touch_summary.join(
    channel_summary["spend"]
)

first_touch_summary["CAC"] = (
    first_touch_summary["spend"] /
    first_touch_summary["attributed_conversions"]
)

first_touch_summary["ROAS"] = (
    first_touch_summary["attributed_revenue"] /
    first_touch_summary["spend"]
)

first_touch_summary.round(2)
```

```python
# Linear Attribution
# Each touchpoint receives equal credit within the customer journey

converted_touchpoints = touchpoints[
    touchpoints["converted"] == 1
].copy()

converted_touchpoints["linear_credit"] = (
    1 / converted_touchpoints["journey_length"]
)

linear_summary = (
    converted_touchpoints
    .groupby("channel")
    .agg(
        attributed_conversions=("linear_credit", "sum"),
        attributed_revenue=("revenue", "sum")
    )
)

linear_summary
```

```python
# Allocate revenue using the same linear attribution credit

converted_touchpoints["linear_attributed_revenue"] = (
    converted_touchpoints["revenue"] *
    converted_touchpoints["linear_credit"]
)

linear_summary = (
    converted_touchpoints
    .groupby("channel")
    .agg(
        attributed_conversions=("linear_credit", "sum"),
        attributed_revenue=("linear_attributed_revenue", "sum")
    )
)

linear_summary = linear_summary.join(
    channel_summary["spend"]
)

linear_summary["CAC"] = (
    linear_summary["spend"] /
    linear_summary["attributed_conversions"]
)

linear_summary["ROAS"] = (
    linear_summary["attributed_revenue"] /
    linear_summary["spend"]
)

linear_summary.round(2)
```

```python
# Convert timestamp to datetime

touchpoints["timestamp"] = pd.to_datetime(
    touchpoints["timestamp"]
)

touchpoints[["user_id", "touch_order", "timestamp"]].head()
```

```python
# Time-Decay Attribution
# More recent touchpoints receive more credit.

converted_touchpoints = touchpoints[
    touchpoints["converted"] == 1
].copy()

# Find the latest touchpoint timestamp for each customer
latest_timestamp = (
    converted_touchpoints
    .groupby("user_id")["timestamp"]
    .transform("max")
)

# Calculate how many days before the final touch each interaction occurred
converted_touchpoints["days_before_conversion"] = (
    latest_timestamp - converted_touchpoints["timestamp"]
).dt.total_seconds() / (60 * 60 * 24)

# Decay factor: credit halves every 7 days
converted_touchpoints["decay_factor"] = (
    0.5 ** (
        converted_touchpoints["days_before_conversion"] / 7
    )
)

# Normalize weights so each customer's total credit = 1
converted_touchpoints["time_decay_credit"] = (
    converted_touchpoints["decay_factor"] /
    converted_touchpoints.groupby("user_id")["decay_factor"].transform("sum")
)

converted_touchpoints[
    [
        "user_id",
        "channel",
        "timestamp",
        "days_before_conversion",
        "time_decay_credit"
    ]
].head(10)
```

```python
# Time-Decay Attribution Summary

converted_touchpoints["time_decay_attributed_revenue"] = (
    converted_touchpoints["revenue"] *
    converted_touchpoints["time_decay_credit"]
)

time_decay_summary = (
    converted_touchpoints
    .groupby("channel")
    .agg(
        attributed_conversions=("time_decay_credit", "sum"),
        attributed_revenue=("time_decay_attributed_revenue", "sum")
    )
)

# Add channel spend
time_decay_summary = time_decay_summary.join(
    channel_summary["spend"]
)

# CAC
time_decay_summary["CAC"] = (
    time_decay_summary["spend"] /
    time_decay_summary["attributed_conversions"]
)

# ROAS
time_decay_summary["ROAS"] = (
    time_decay_summary["attributed_revenue"] /
    time_decay_summary["spend"]
)

time_decay_summary.round(2)
```

```python
# Position-Based Attribution

converted_users = touchpoints[
    touchpoints["converted"] == 1
].copy()

converted_users = converted_users.sort_values(
    ["user_id", "touch_order"]
)

def position_based_credit(group):
    n = len(group)

    if n == 1:
        credits = [1.0]

    elif n == 2:
        credits = [0.5, 0.5]

    else:
        middle_credit = 0.20 / (n - 2)
        credits = [0.40] + [middle_credit] * (n - 2) + [0.40]

    group = group.copy()
    group["position_based_credit"] = credits

    return group


position_based_touchpoints = (
    converted_users
    .groupby("user_id", group_keys=False)
    .apply(position_based_credit)
    .reset_index(drop=True)
)

position_based_touchpoints.head(10)
```

```python
# Position-Based Attribution Summary

position_based_touchpoints["position_based_attributed_revenue"] = (
    position_based_touchpoints["revenue"] *
    position_based_touchpoints["position_based_credit"]
)

position_based_summary = (
    position_based_touchpoints
    .groupby("channel")
    .agg(
        attributed_conversions=("position_based_credit", "sum"),
        attributed_revenue=("position_based_attributed_revenue", "sum")
    )
)

# Add channel spend
position_based_summary = position_based_summary.join(
    channel_summary["spend"]
)

# CAC
position_based_summary["CAC"] = (
    position_based_summary["spend"] /
    position_based_summary["attributed_conversions"]
)

# ROAS
position_based_summary["ROAS"] = (
    position_based_summary["attributed_revenue"] /
    position_based_summary["spend"]
)

position_based_summary.round(2)
```

```python
# Compare attributed conversions across all models

attribution_comparison = pd.concat(
    {
        "First-Touch": first_touch_summary["attributed_conversions"],
        "Last-Touch": last_touch_summary["attributed_conversions"],
        "Linear": linear_summary["attributed_conversions"],
        "Time-Decay": time_decay_summary["attributed_conversions"],
        "Position-Based": position_based_summary["attributed_conversions"]
    },
    axis=1
)

attribution_comparison.round(2)
```

```python
# Inspect benchmark-related logic in the data generator

with open("data/raw/generate_attribution_data.py", "r", encoding="utf-8") as f:
    generator_code = f.read()

for line in generator_code.splitlines():
    if "ground_truth" in line.lower() or "benchmark" in line.lower():
        print(line)
```

```python
# Create the controlled synthetic benchmark

touchpoints["role"] = np.where(
    touchpoints["journey_length"] == 1,
    "closer",
    np.where(
        touchpoints["touch_order"] == 1,
        "opener",
        np.where(
            touchpoints["touch_order"] == touchpoints["journey_length"],
            "closer",
            "middle"
        )
    )
)

CHANNEL_PROFILE = {
    "Paid Social":    {"opener": 0.90, "closer": 0.20, "middle": 0.40},
    "Display":        {"opener": 0.70, "closer": 0.15, "middle": 0.30},
    "Organic Search": {"opener": 0.50, "closer": 0.50, "middle": 0.40},
    "Referral":       {"opener": 0.50, "closer": 0.40, "middle": 0.40},
    "Paid Search":    {"opener": 0.30, "closer": 0.80, "middle": 0.40},
    "Email":          {"opener": 0.10, "closer": 0.70, "middle": 0.50},
    "Direct":         {"opener": 0.20, "closer": 0.90, "middle": 0.30},
}

# Assign the benchmark weight to each touchpoint
touchpoints["ground_truth_weight"] = touchpoints.apply(
    lambda row: CHANNEL_PROFILE[row["channel"]][row["role"]],
    axis=1
)

# Only converted journeys are evaluated
ground_truth_data = touchpoints[
    touchpoints["converted"] == 1
].copy()

# Normalize benchmark weights within each customer journey
ground_truth_data["ground_truth_credit"] = (
    ground_truth_data["ground_truth_weight"] /
    ground_truth_data.groupby("user_id")["ground_truth_weight"].transform("sum")
)

# Allocate revenue using benchmark credit
ground_truth_data["ground_truth_attributed_revenue"] = (
    ground_truth_data["revenue"] *
    ground_truth_data["ground_truth_credit"]
)

ground_truth_summary = (
    ground_truth_data
    .groupby("channel")
    .agg(
        attributed_conversions=("ground_truth_credit", "sum"),
        attributed_revenue=("ground_truth_attributed_revenue", "sum")
    )
)

ground_truth_summary.round(2)
```

```python
# Compare all attribution models against the controlled benchmark

model_comparison = pd.concat(
    {
        "First-Touch": first_touch_summary["attributed_conversions"],
        "Last-Touch": last_touch_summary["attributed_conversions"],
        "Linear": linear_summary["attributed_conversions"],
        "Time-Decay": time_decay_summary["attributed_conversions"],
        "Position-Based": position_based_summary["attributed_conversions"],
        "Ground Truth": ground_truth_summary["attributed_conversions"]
    },
    axis=1
)

model_comparison = model_comparison.round(2)

model_comparison
```

```python
# Calculate attribution error for each model

benchmark = model_comparison["Ground Truth"]

mae_results = {}

for model in [
    "First-Touch",
    "Last-Touch",
    "Linear",
    "Time-Decay",
    "Position-Based"
]:
    mae_results[model] = (
        model_comparison[model] - benchmark
    ).abs().mean()

mae_table = pd.DataFrame(
    mae_results.items(),
    columns=["Model", "MAE"]
).sort_values("MAE")

mae_table.round(2)
```

## Final Business Insights

### 1. Customer Journey Behavior
Longer customer journeys show higher conversion rates, indicating that customers with multiple marketing interactions are more likely to convert.

### 2. Attribution Model Impact
Attribution model choice significantly changes the amount of conversion credit assigned to each marketing channel.

### 3. Last-Touch Limitation
Last-Touch attribution heavily favors channels that appear near the end of the customer journey and can undervalue channels that contribute earlier in the journey.

### 4. Paid Channel Evaluation
The three paid channels — Paid Search, Paid Social, and Display — should be evaluated using multi-touch attribution rather than Last-Touch alone because their contribution changes substantially across models.

### 5. Benchmark Validation
Against the controlled synthetic benchmark, Position-Based Attribution achieved the lowest Mean Absolute Error (MAE), followed by Time-Decay and Linear Attribution.

### 6. Business Recommendation
For budget allocation decisions, management should not rely exclusively on Last-Touch attribution. A multi-touch approach, particularly Position-Based Attribution in this controlled analysis, provides a more balanced view of channel contribution.

### Important Limitation
The benchmark represents the known generative structure of this synthetic dataset. It is a validation benchmark, not real-world causal ground truth.
